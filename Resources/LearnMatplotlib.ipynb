## Subplots and legends

```python
# Trading Days
days = [1, 2, 3, 4, 5]
original_price = [10, 20, 25, 30, 40]
predicted_price = [8, 21, 25, 28, 38]
```

```python
plt.plot(days, original_price, 'b', label= 'Original price')    # Blue line - Original price
plt.plot(days, predicted_price, 'r', label= 'Predicted price')  # Red line - Predicted price
plt.legend()
plt.title('Plot with Legends')
plt.xlabel('Weekdays')
plt.ylabel('Price')
```

```python
# Add Data Points
plt.plot(days, original_price, 'b-o', label= 'Original price')    # Blue line - Original price
plt.plot(days, predicted_price, 'r-o', label= 'Predicted price')  # Red line - Predicted price
plt.legend()
plt.title('Plot with Legends')
plt.xlabel('Weekdays')
plt.ylabel('Price')
```

### Subplot

```python
# First subplot - Original price
plt.subplot(3, 1, 1)                     # (no. of rows/grid , no.of cloumns, index )
plt.plot(days, original_price, 'b-o', label= 'Original price')
plt.title('Original prices over time')
plt.xlabel('Weekdays')
plt.ylabel('Price')
plt.legend()


# Second subplot - Predicted price
plt.subplot(3, 1, 2)                      # (no. of rows/grid , no.of cloumns, index )
plt.plot(days, predicted_price, 'r-o', label= 'Predicted price')
plt.title('Predicted prices over time')
plt.xlabel('Weekdays')
plt.ylabel('Price')
plt.legend()


# Third subplot - DIfferents between Original & Predicted prices
price_difference = [o - p for o, p in zip(original_price, predicted_price) ]

plt.subplot(3, 1, 3)                      # (no. of rows/grid , no.of cloumns, index )
plt.bar(days, price_difference, color='purple', label= 'Price difference')
plt.title('DIfferents between Original & Predicted prices')
plt.xlabel('Days')
plt.ylabel('Price Difference')
plt.legend()


plt.tight_layout()    # To avoid padding , adding Padding


# Save the Image
plt.savefig('Example_Prediction_&_difference.png', dpi=300)  
```

### Saving the figure
